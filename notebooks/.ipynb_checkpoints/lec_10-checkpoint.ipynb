## Running the extended Euclidean algorithm


To run the EA to verify that $\gcd(m(x),p(x)) = 1$ for the polynomials
$$m(x) = x^8 +x^4 + x^3 + x +1    $$
and 
$$p(x) = x^5 + x^3 + x^2 + 1$$.

The first polynomial division with remainder is 
$$m(x) = (x^3 + x + 1) p(x) + x^2.$$

The second division is then 
$$ p(x) = (x^3 + x + 1)x^2 + 1.$$

From this we see that the greatest common divisor is 1 as expected.

Find the multiplicative inverse $p^{-1}(x)$ we use back substitution on the polynomial divisions giving the resulting Bezout's equation of 

$$1 = (x^3 + x +1 )m(x) + (x^6 + x^2 )p(x).$$

From this we see that 
$$p^{-1}(x) = x^6 + x^2.$$
